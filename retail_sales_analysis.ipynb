```python

import pandas as pd

data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Laptop", "Mouse",
                "Monitor", "Keyboard", "Monitor"],
    "Category": ["Electronics", "Accessories", "Accessories",
                 "Electronics", "Accessories", "Electronics",
                 "Accessories", "Electronics"],
    "Quantity": [2, 10, 5, 1, 15, 3, 8, 2],
    "Price": [50000, 500, 1500, 50000, 500, 12000, 1500, 12000]
}

df = pd.DataFrame(data)
df["Revenue"] = df["Quantity"] * df["Price"]

df
```

```python
from numpy import average
total_revenue = df["Revenue"].sum()
total_units = df["Quantity"].sum()
average_revenue = df["Revenue"].mean()
print("Total Revenue:", total_revenue )
print("Total Units", total_units)
print("Average Revenue per Product Entry:", average_revenue)
```

```python
Product_sales = (
    df.groupby("Product")["Revenue"]
    .sum()
    .sort_values(ascending=False)

)
print(Product_sales)
```

```python
import matplotlib.pyplot as plt
Product_sales.plot(kind="bar", title="Revenue by Product")

plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd

data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Laptop", "Mouse",
                "Monitor", "Keyboard", "Monitor"],
    "Category": ["Electronics", "Accessories", "Accessories",
                 "Electronics", "Accessories", "Electronics",
                 "Accessories", "Electronics"],
    "Quantity": [2, 10, 5, 1, 15, 3, 8, 2],
    "Price": [50000, 500, 1500, 50000, 500, 12000, 1500, 12000]
}

df = pd.DataFrame(data)
df["Revenue"] = df["Quantity"] * df["Price"]

product_sales = (
    df.groupby("Product")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

print(product_sales)
```

```python

best_product = product_sales.idxmax()
best_revenue = product_sales.max()

print("Best product:", best_product)
print("Total revenue:", best_revenue)
```

```python

category_sales = (
    df.groupby("Category")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

print(category_sales)
```

## Business Insights — Retail Sales Analysis

1. **Highest-revenue product:** Laptop generated the highest total revenue of ₹1,50,000, making it the leading product by revenue in this sample dataset.

2. **Category performance:** Electronics generated ₹3,48,000 in revenue, compared with ₹41,000 from Accessories. Electronics contributed the larger share of total revenue.

3. **Business recommendation:** The business should investigate demand for electronic products and maintain appropriate stock levels. However, it should also compare product costs and profit margins before deciding which products deserve greater investment.
